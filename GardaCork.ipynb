```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def getCSV(name):
    dfl = pd.read_csv(name)
    return dfl
df = getCSV(r'C:\Users\Tom\OneDrive\Documents\APROJECTS\GardaCork.csv') 
print(list(set(df['Type of Offence'])))
```

```python
county = "Cork City Garda Division"
offence = "Offences while in custody, breach of court orders"
filtered_df = df[df['Garda Division']==county]
filtered_df = filtered_df[filtered_df['Type of Offence']==offence]
print(filtered_df)
```

```python
def changeQuarter(val):
    year = val[:4]
    quarter =  val[-2:]
    return str(quarter)+" "+str(year)
filtered_df['Quarter'] = filtered_df['Quarter'].apply(changeQuarter)
```

```python
fig, ax = plt.subplots(1, 1, figsize=(12, 8))

filtered_df.plot(x="Quarter", y="VALUE", kind="line", marker="None", linestyle="-", linewidth=2, color="#D33", ax=ax, label=offence)

title_string = "Crime Incidents Reported to Cork City Garda Division, per quarter, Q1 2003 - Q1 2025."

plt.title(title_string)
plt.xlabel('Year', fontsize=14)
plt.ylabel('Incidents Reported',fontsize=14)
ax.set_facecolor('#EEEEEE') 
plt.grid(True)

plt.legend(fontsize=14)

plt.text(0.75, -0.12, "Data Source: https://data.cso.ie/\n\"CJQ06 Recorded crime incidents\" ", ha='left',
         transform=ax.transAxes, fontdict={'fontstyle': 'italic', 'fontsize': 10})

save_fig = 'Cork-'+offence+'.png'
plt.savefig(save_fig)

# Show the plot
plt.show()
```
