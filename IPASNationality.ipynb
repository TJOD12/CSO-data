```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
```

```python
def getCSV(csv):
    dfl = pd.read_csv(csv)
    return dfl
df = getCSV(r'C:\Users\Tom\OneDrive\Documents\APROJECTS\IPASNationality.csv') 
print(df)
print(list(set(df["Country of Nationality"])))
```

```python
def changeNames(name):
    switch = {
        "Congo (the Democratic Republic of the)": "DR Congo",
        "Palestine, State of": "Palestine",
        "Syrian Arab Republic (the)": "Syria",
        "Sudan (the)": "Sudan",
        "Africa(3)": "Other Africa",
        "Asia(2)": "Other Asia",
        "America(3)": "Other America"
    }
    if name not in switch:
        return name 
    else:
        return switch[name]
```

```python
old_year = 2016
new_year = 2024
filtered_df = df[df["Country of Nationality"]!="All citizenships"]
filtered_df = filtered_df[filtered_df["Country of Nationality"]!="Selected European Countries(2)"]
old_df = filtered_df[filtered_df["Year Daily Expense Allowance Activity Started"]==old_year]
new_df = filtered_df[filtered_df["Year Daily Expense Allowance Activity Started"]==new_year]

old_df["Country of Nationality"] = old_df["Country of Nationality"].apply(changeNames)
new_df["Country of Nationality"] = new_df["Country of Nationality"].apply(changeNames)
print(new_df)
```

```python
fig, ax = plt.subplots(1, 1, figsize=(12, 8))

new_df.plot(x="Country of Nationality", y="VALUE", kind="barh", color="#22BB33", ax=ax, label="2024")
old_df.plot(x="Country of Nationality", y="VALUE", kind="barh", color="#AA2233", ax=ax, label="2016")

title_string = "Growth of International Protection Applicants to Ireland, 2016 vs 2024."

plt.title(title_string)
plt.xlabel('Applicants', fontsize=14)
plt.ylabel('')
ax.set_ylim(0)
ax.set_facecolor('#EEEEEE') 
plt.grid(True, axis="x")

plt.legend(fontsize=15)

plt.text(0.65, -0.135, "Data Source: https://data.cso.ie/\n\"IAIP02 International protection applicants \nidentified using administrative data\" ", ha='left',
         transform=ax.transAxes, fontdict={'fontstyle': 'italic', 'fontsize': 10})

save_fig = 'IPAS-2016VS2024.png'
plt.savefig(save_fig, bbox_inches="tight")

# Show the plot
plt.show()
```

```python
age_df = getCSV(r'C:\Users\Tom\OneDrive\Documents\APROJECTS\IPASAge.csv') 
```

```python
filtered_age_df = age_df[age_df['Sex']=="Both sexes"]
filtered_age_df = filtered_age_df[filtered_age_df['Age Group']!="All ages"]
print(filtered_age_df)
```

```python
fig, ax = plt.subplots(1,1,figsize=(12,8))

pivot_df = filtered_age_df.pivot_table(index='Year Daily Expense Allowance Activity Started', columns='Age Group', values='VALUE', aggfunc='sum', fill_value=0)

desired_order = [
    "0 - 14 years",
    "15 - 24 years",
    "25 - 44 years",
    "45 years and over"
]

pivot_df = pivot_df.reindex(columns=desired_order)

colors = ["#FCBB7C", "#FF963D", "#E06F09", "#B85A04"]
pivot_df.plot(kind='bar', stacked=True, figsize=(12, 9), ax=ax, color=colors, width= .5)

ax.set_facecolor('#EEEEEE') 
# plt.grid(True)
plt.xlabel('Year', fontsize=14)
plt.ylabel('Applicants', fontsize=14)
plt.grid(axis='y')
plt.legend(loc='upper left')

plt.xticks(rotation=0)
plt.title('Age Breakdown of International Protection Applicants to Ireland, 2016 - 2024.')

plt.subplots_adjust(left=0.08, right=0.8, top=0.9, bottom=0.15)
plt.text(0.65, -0.125, "Data Source: https://data.cso.ie/ \n\"IAIP02 International protection applicants \nidentified using administrative data\"", ha='left', transform=ax.transAxes)

plt.savefig('IPAS-Age-2016-2024.png', bbox_inches='tight')

plt.show()
```
