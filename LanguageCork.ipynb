```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def getCSV(name):
    dfl = pd.read_csv(name)
    return dfl
df = getCSV(r'C:\Users\Tom\OneDrive\Documents\APROJECTS\LanguageCork.csv') 
print(df)
```

```python
county =  "Dublin"
cork_df =  df[df['CSO Local Electoral Areas 2022'].str.contains(county)]
print(set(list(cork_df['CSO Local Electoral Areas 2022'])))
```

```python
town = "Dublin"
filtered_df =  df[df['CSO Local Electoral Areas 2022'].str.contains(town)]
filtered_df = filtered_df[filtered_df['Language']=="Total"]
print(filtered_df)
```

```python
pop_df = getCSV(r'C:\Users\Tom\OneDrive\Documents\APROJECTS\CorkPopulation2022.csv') 
pop_county = "Dublin"
total_pop_df = pop_df[(pop_df['Age']=="Total")&(pop_df['Sex']=="Both Sexes")]
total_pop_df =  total_pop_df[total_pop_df['CSO Local Electoral Areas 2022'].str.contains(pop_county)]
print(total_pop_df)
```

```python
total_pop_df['CSO Local Electoral Areas 2022'] = total_pop_df['CSO Local Electoral Areas 2022'].apply(lambda x: x.split(",")[0])
filtered_df['CSO Local Electoral Areas 2022'] = filtered_df['CSO Local Electoral Areas 2022'].apply(lambda x: x.split(",")[0])
```

```python
fig, ax = plt.subplots(1, 1, figsize=(12, 8))

total_pop_df.plot(x="CSO Local Electoral Areas 2022", y="VALUE", kind="barh", color="#227", ax=ax, label="Total Population")
filtered_df.plot(x="CSO Local Electoral Areas 2022", y="VALUE", kind="barh", color="#DA3", ax=ax, label="Foreign Speakers")

title_string = "Speakers of Foreign Languages within Dublin CSO Local Electoral Areas, 2022."

plt.title(title_string)
plt.xlabel('Population', fontsize=14)
plt.ylabel('Electoral Area',fontsize=14)
ax.set_facecolor('#EEEEEE') 
plt.grid(True, axis="x")

plt.legend(fontsize=12)

plt.text(0.75, -0.12, "Data Source: https://data.cso.ie/\n\"SAP2022T1T1LEA22 Population\" ", ha='left',
         transform=ax.transAxes, fontdict={'fontstyle': 'italic', 'fontsize': 10})

save_fig = 'Dublin-Foreign-Language.png'
plt.savefig(save_fig, bbox_inches='tight')

# Show the plot
plt.show()
```
