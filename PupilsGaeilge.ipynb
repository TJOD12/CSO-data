```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
```

```python
def getCSV(csv):
    dfl = pd.read_csv(csv)
    return dfl
df = getCSV(r'C:\Users\Tom\OneDrive\Documents\APROJECTS\PupilsGaeilge.csv') 
print(df)
print(list(set(df["Type of School"])))
```

```python
label = "All Irish Medium Schools"
filtered_df = df[df["Type of School"]==label]
filtered_df = filtered_df[filtered_df["County"]!="All Counties"]
filtered_df = filtered_df[filtered_df["Academic Year"]=="2023-2024"]
```

```python
fig, ax = plt.subplots(1, 1, figsize=(13, 6))

filtered_df.plot(x="County", y="VALUE", kind="barh", color="#1D984A", ax=ax, label="Pupils")

ax.set_title(label+", 2023-2024.", fontsize=14)
ax.set_xlabel("Number of Schools", fontsize=13)
ax.set_ylabel("")
#ax.set_ylim(0, 31500)
ax.tick_params(axis='x', rotation=0)
ax.set_facecolor('#EEEEEE')
ax.get_legend().remove()
ax.grid(axis='x')

plt.text(0.74, -0.18, "Data Source: https://data.cso.ie/\n\"ED111 Primary School Pupils who \nare taught through Irish\" ", ha='left',
         transform=ax.transAxes, fontdict={'fontstyle': 'italic', 'fontsize': 10})

save_fig = 'gaeilgePupils'+label+'.png'
plt.savefig(save_fig, bbox_inches="tight")

plt.show()
```

```python
foreign_df = getCSV(r'C:\Users\Tom\OneDrive\Documents\APROJECTS\NationalityPupilsCompare.csv') 
print(foreign_df)
```

```python
f_filtered_df = foreign_df[foreign_df["County"]!="All Counties"]
f_filtered_df = f_filtered_df[f_filtered_df["Academic Year"]!="2016-2017"]
f_filtered_df = f_filtered_df[f_filtered_df["Nationality of Pupil"]!="Ireland"]
f_filtered_df = f_filtered_df[f_filtered_df["Nationality of Pupil"]!="All countries"]
vale = f_filtered_df.groupby('Academic Year')['VALUE'].sum()
df_vale = vale.reset_index()
print(df_vale)
```

```python
fig, ax = plt.subplots(1, 1, figsize=(10, 6))

df_vale.plot(x="Academic Year", y="VALUE", kind="line", linestyle="-", linewidth=3, color="red", ax=ax, label="Foreign pupils")

title_string = "Foreign National Pupils Attending Primary School, Ireland 2016 - 2024"

plt.title(title_string)
plt.xlabel('Academic Year', fontsize=14)
plt.ylabel('Pupils', fontsize=14)
ax.set_facecolor('#EEEEEE') 
ax.get_legend().remove()
plt.grid(True)

plt.text(0.74, -0.18, "Data Source: https://data.cso.ie/\n\"ED123Nationality of Pupils \nattending Primary School\" ", ha='left',
         transform=ax.transAxes, fontdict={'fontstyle': 'italic', 'fontsize': 10})

plt.savefig('Pupils-Foreign-line.png')

# Show the plot
plt.show()
```

```python
def tidy(year):
    y1 = year[2:4]
    gap = year[4:5]
    y2 = year[7:9]
    return y1+gap+y2
df_vale["Academic Year"] = df_vale["Academic Year"].apply(tidy)
filtered_df["Academic Year"] = filtered_df["Academic Year"].apply(tidy)
print(df_vale)
```

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 6))

df_vale.plot(x="Academic Year", y="VALUE", kind="line", linestyle="-", linewidth=3, color="red", ax=ax[0], label="Foreign pupils")
filtered_df.plot(x="Academic Year", y="VALUE", kind="line", linestyle="--", linewidth=3.3, color="green", ax=ax[1], label="Irish Medium School Pupils")

ax[0].set_title("Foreign National Pupils Attending Primary School")
ax[0].set_xlabel("Academic Year", fontsize=14)
ax[0].set_ylabel("Number of Pupils")
#ax[0].set_ylim(0, 31500)
ax[0].tick_params(axis='x', rotation=0)
ax[0].set_facecolor('#EEEEEE')
ax[0].grid(axis='y')
plt.text(0.67, -0.22, "Data Source: https://data.cso.ie/\n\"ED111 Primary School Pupils who \nare taught through Irish\" ", ha='left',
         transform=ax[1].transAxes, fontdict={'fontstyle': 'italic', 'fontsize': 10})

ax[1].set_title("All Mainstream Pupils in Irish Language Medium Schools")
ax[1].set_xlabel("Academic Year", fontsize=14)
ax[1].set_ylabel("Number of Pupils")
#ax[1].set_ylim(0, 31500)
ax[1].tick_params(axis='x', rotation=0)
ax[1].set_facecolor('#EEEEEE')
ax[1].grid(axis='y')
plt.text(0.67, -0.22, "Data Source: https://data.cso.ie/\n\"ED123 Nationality of Pupils \nattending Primary School\" ", ha='left',
         transform=ax[0].transAxes, fontdict={'fontstyle': 'italic', 'fontsize': 10})

save_fig = 'Pupils-Gaeilge-Foreign-line.png'
plt.savefig(save_fig, bbox_inches="tight")

# Show the plot
plt.show()
```
