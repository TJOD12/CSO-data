```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
```

```python
def getCSV(csv):
    dfl = pd.read_csv(csv)
    return dfl
df = getCSV(r'C:\Users\Tom\OneDrive\Documents\APROJECTS\RailThings.csv') 
print(df)
```

```python
years = [1982,1992,2002,2012,2022]
filtered_df = df[df['Year'].isin(years)]
filtered_df = filtered_df[~(filtered_df['Commodity'] == "All commodities")]
```

```python
fig, ax = plt.subplots(1,1,figsize=(12,8))

colors = ["#A10000", "#D30973","#888888","#D5D97E","#5D58AD","#666666","#005F6A","#BA857D"]
pivot_df = filtered_df.pivot_table(index='Year', columns='Commodity', values='VALUE', aggfunc='sum', fill_value=0)

pivot_df.plot(kind='bar', stacked=True, figsize=(12, 8), ax=ax, color=colors,edgecolor='black')

ax.set_facecolor('#EEEEEE') 
# plt.grid(True)
plt.xlabel('Year')
plt.ylabel('Tonnes')
plt.xticks(rotation=0)
plt.title('Principal Commodities Conveyed by Rail, 1982 - 2022. (tonnes)')

plt.subplots_adjust(left=0.1, right=0.9, top=0.9, bottom=0.15)
plt.text(0.64, -0.125, "Data Source: https://data.cso.ie/ \n\"TCA03 Principal Commodities Conveyed by Rail\"", ha='left', transform=ax.transAxes)

plt.savefig("RailCommodities1982.png")

plt.show()
```
