```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
def getCSV(csv):
    dfl = pd.read_csv(csv)
    return dfl
df = getCSV(r'C:\Users\Tom\OneDrive\Documents\APROJECTS\RegionsAge.csv') 
print(df)
```

```python
year=2024
filtered_df = df[df['Age Group']!="All ages"]
filtered_df = filtered_df[filtered_df['Year']==year]
filtered_df = filtered_df[(filtered_df['Sex']=='Both sexes')&(filtered_df['NUTS 3 Region']!='Ireland')]
print(filtered_df)
```

```python
region="Midland"
filtered_df = filtered_df[filtered_df['NUTS 3 Region']==region]
print(filtered_df)
```

```python
fig, ax = plt.subplots(1, 1, figsize=(13, 8))

filtered_df.plot(kind='barh', x='Age Group', y='VALUE', color="#386", label="", width=0.8, ax=ax)

title_string = "Estimated population of "+region+" region by age group, "+str(year)+"."

plt.title(title_string, fontsize=15)
plt.xlabel('Population', fontsize=14)
plt.ylabel('Age Group', fontsize=14)
plt.xticks(rotation=0) 
#ax.set_xlim(200,550)
ax.set_facecolor('#EEEEEE') 
ax.legend(fontsize=14) 
plt.grid(True, axis="x")


plt.subplots_adjust(left=0.1, right=0.9, top=0.9, bottom=0.17)
plt.text(0.75, -0.13, "Data Source: https://data.cso.ie/ \n\"PEA26 Estimated Population \n(Persons in January)\"", ha='left', transform=ax.transAxes)

save_fig = 'RegionAge-'+region+'.png'
plt.savefig(save_fig)

# Show the plot
plt.show()
```
