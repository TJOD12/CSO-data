```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def getCSV(name):
    dfl = pd.read_csv(name)
    return dfl
df = getCSV(r'C:\Users\Tom\OneDrive\Documents\APROJECTS\RentCork.csv') 
print(df)
```

```python
copy_df = df[df['Location'].str.contains('Cork')]
print(list(set(copy_df['Location'])))
```

```python
town = "Fermoy"
semi = 'Semi detached house'
apartment = 'Apartment'
filtered_df = df[df['Location'].str.contains(town)]
filtered_df = filtered_df[filtered_df['Number of Bedrooms']=='All bedrooms']
semi_df = filtered_df[filtered_df['Property Type']==semi]
apartment_df = filtered_df[filtered_df['Property Type']==apartment]
```

```python
print(filtered_df)
```

```python
fig, ax = plt.subplots(1, 1, figsize=(12, 8))

semi_df.plot(x="Year", y="VALUE", kind="line", marker="None", linestyle="-", linewidth=3, color="#C55", ax=ax, label=semi)
apartment_df.plot(x="Year", y="VALUE", kind="line", marker="None", linestyle="-", linewidth=3, color="#55C", ax=ax, label=apartment)

title_string = "Average Monthly Rent Report, "+town+", Cork, 2008 - 2024."

plt.title(title_string)
plt.xlabel('Year', fontsize=14)
plt.ylabel('Monthly Rent (Є)',fontsize=14)
ax.set_facecolor('#EEEEEE') 
plt.grid(True)

plt.legend(fontsize=14)

plt.text(0.75, -0.12, "Data Source: https://data.cso.ie/\n\"RIA02 RTB Average Monthly Rent Report\" ", ha='left',
         transform=ax.transAxes, fontdict={'fontstyle': 'italic', 'fontsize': 10})

save_fig = 'Cork-'+semi+apartment+town+'.png'
plt.savefig(save_fig)

# Show the plot
plt.show()
```
