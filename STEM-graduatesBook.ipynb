```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
```

```python
def getCSV(csv):
    dfl = pd.read_csv(csv)
    return dfl

df = getCSV(r'C:\Users\Tom\OneDrive\Documents\APROJECTS\STEM_graduates_europe.csv')
print(df)
```

```python
filtered_df = df[(df['Year'] == 2021)&(df['Sex']=='Total Persons')]
filtered_df = filtered_df[~(filtered_df['Countries']=='United Kingdom')]
print(filtered_df)
```

```python
fig, ax = plt.subplots(1,1,figsize=(12,9))

colors=["grey" for i in range(1,27)]
colors[15] = "#47C96C"

plt.barh(filtered_df['Countries'], filtered_df['VALUE'], color=colors)

title_string = 'STEM graduates per 1000 of population aged 20 - 29, 2021.'

plt.title(title_string)
plt.xlabel('Number of graduates (per 1000)')
plt.ylabel('')
plt.xticks(rotation=0)
#plt.legend(title='Area', loc='upper left', label="None", fontsize='11')
plt.legend().remove()
ax.set_facecolor('#EEEEEE') 
plt.tight_layout()

plt.subplots_adjust(left=0.2, right=0.9, top=0.9, bottom=0.15)
plt.text(0.71, -0.115, "Data Source: https://data.cso.ie/ \n\"MIP09 Science, Technology, Engineering \nand Mathematics graduates\"", ha='left', transform=ax.transAxes)

plt.savefig('STEM-graduates-Europe-2021.png')

plt.show()
```
