```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
```

```python
def getCSV(csv):
    dfl = pd.read_csv(csv)
    return dfl
df = getCSV(r'C:\Users\Tom\OneDrive\Documents\APROJECTS\suicideRates.csv') 
print(df)
```

```python
filtered_df = df[df['Statistic Label']=="Deaths"]
male_df = filtered_df[filtered_df['Sex']=="Male"]
female_df = filtered_df[filtered_df['Sex']=="Female"]
```

```python
fig, ax = plt.subplots(1,1,figsize=(10,6))

male_df.plot(x="Year", y="VALUE", marker='None', color='#239', linestyle='-', linewidth=2.5, markersize=5, ax=ax, label="Males")
female_df.plot(x="Year", y="VALUE", marker='None', color='#F28', linestyle='-', linewidth=2.5, markersize=5, ax=ax, label="Females")

title_string = "Number of Suicides per year, 2000 - 2024."
ax.set_title(title_string)
ax.set_facecolor('#EEEEEE') 
#ax.set_ylim(110, 240)
plt.title(label=title_string, fontsize=14)
plt.xlabel('Year', fontsize=13)
plt.ylabel('Number of Suicides', fontsize=13)
plt.legend(title='', loc='upper right', fontsize='11')
plt.tight_layout()
plt.grid(True)

plt.subplots_adjust(left=0.1, right=0.90, top=0.90, bottom=0.15)
plt.text(0.7, -0.14, "Data Source: https://data.cso.ie/ \n\"VSD32 Suicide death rates\"", ha='left', transform=ax.transAxes)

plt.savefig("SuicideRates2000.png")

plt.show()
```
