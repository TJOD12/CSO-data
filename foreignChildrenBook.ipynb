```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
```

```python
def getCSV(csv):
    dfl = pd.read_csv(csv)
    return dfl
df = getCSV(r'C:\Users\Tom\OneDrive\Documents\APROJECTS\foreignChildren.csv') 
print(df)
```

```python
year = 2022
filtered_df = df[df['Year']==year]
filtered_df = filtered_df[filtered_df['Administrative County']!="All Counties"]

print(filtered_df)
```

```python
def changeNames(name):
    switch = {
        "Dún Laoghaire-Rathdown":"DLR"
    }
    
    if name in switch:
        return switch[name]
    else: 
        return name
    
filtered_df['Administrative County'] = filtered_df['Administrative County'].apply(changeNames)
```

```python
fig, ax = plt.subplots(1, 1, figsize=(14, 8))

filtered_df.plot(x="Administrative County", y="VALUE", kind="barh", width=0.75, edgecolor="black", color="#929", ax=ax)

label = "Number of Foreign National Children per County"
title_string = label+", " + str(year) + "."

plt.title(title_string)
plt.xlabel('Children', fontsize=14)
plt.ylabel('County', fontsize=14)
plt.xticks(rotation=0) 
#ax.set_xlim(200,550)
ax.set_facecolor('#EEEEEE') 
ax.get_legend().remove()
plt.grid(True, axis="x")

plt.subplots_adjust(left=0.1, right=0.9, top=0.9, bottom=0.17)
plt.text(0.7, -0.15, "Data Source: https://data.cso.ie/ \n\"SCA41 Number of Foreign National Children\"", ha='left', transform=ax.transAxes)

save_fig = 'foreignChildren.png'
plt.savefig(save_fig)

# Show the plot
plt.show()
```
